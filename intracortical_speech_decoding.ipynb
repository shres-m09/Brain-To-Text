```python
import h5py

file_path = "/content/brain-25-data/data_train.hdf5"

with h5py.File(file_path, "r") as f:
    print("Number of trials:", len(f.keys()))
    print("First 5 trial IDs:", list(f.keys())[:5])
```

```python
with h5py.File(file_path, "r") as f:
    first_key = list(f.keys())[0]
    trial = f[first_key]

    print("Trial ID:", first_key)
    print("\nData stored in this trial:")
    print(list(trial.keys()))

    print("\nTrial attributes:")
    for key, value in trial.attrs.items():
        print(key, ":", value)
```

```python
with h5py.File(file_path, "r") as f:
    trial = f["trial_0000"]

    neural = trial["input_features"][:]
    phonemes = trial["seq_class_ids"][:]
    transcription = trial["transcription"][:]

    print("Neural shape:", neural.shape)
    print("Neural dtype:", neural.dtype)

    print("\nPhoneme IDs:", phonemes)
    print("Phoneme shape:", phonemes.shape)

    print("\nTranscription:", transcription)
```

```python
import numpy as np

with h5py.File(file_path, "r") as f:
    trial = f["trial_0000"]

    neural = trial["input_features"][:]
    phonemes = trial["seq_class_ids"][:]

print("First 5 time steps:")
print(neural[:5, :10])

print("\nFeature means:")
print(neural.mean(axis=0)[:10])

print("\nFeature standard deviations:")
print(neural.std(axis=0)[:10])

print("\nPhoneme IDs:")
print(phonemes)
```

```python
with h5py.File(file_path, "r") as f:
    trial = f["trial_0000"]
    neural = trial["input_features"][:]

print("Total features:", neural.shape[1])

# Check whether the two halves have the same number of features
print("First half:", neural[:, :256].shape)
print("Second half:", neural[:, 256:].shape)
```

```python
with h5py.File(file_path, "r") as f:

    time_lengths = []
    sentence_lengths = []

    for trial_id in f.keys():
        trial = f[trial_id]

        time_lengths.append(trial.attrs["n_time_steps"])
        sentence_lengths.append(trial.attrs["seq_len"])

print("Number of trials:", len(time_lengths))

print("Time steps:")
print("  Minimum:", min(time_lengths))
print("  Maximum:", max(time_lengths))
print("  Average:", sum(time_lengths) / len(time_lengths))

print("\nPhoneme sequence lengths:")
print("  Minimum:", min(sentence_lengths))
print("  Maximum:", max(sentence_lengths))
print("  Average:", sum(sentence_lengths) / len(sentence_lengths))
```

```python
# Let's inspect the label mapping for trial 0

LOGIT_TO_PHONEME = [
    'BLANK',
    'AA', 'AE', 'AH', 'AO', 'AW',
    'AY', 'B', 'CH', 'D', 'DH',
    'EH', 'ER', 'EY', 'F', 'G',
    'HH', 'IH', 'IY', 'JH', 'K',
    'L', 'M', 'N', 'NG', 'OW',
    'OY', 'P', 'R', 'S', 'SH',
    'T', 'TH', 'UH', 'UW', 'V',
    'W', 'Y', 'Z', 'ZH',
    ' | '
]

with h5py.File(file_path, "r") as f:
    trial = f["trial_0000"]
    ids = trial["seq_class_ids"][:]
    seq_len = trial.attrs["seq_len"]

actual_ids = ids[:seq_len]

print("Phoneme IDs:", actual_ids)
print("Phonemes:", [LOGIT_TO_PHONEME[i] for i in actual_ids])
```
